```python
# 1. Applying the random Forest Regressor on the ames dataset

from sklearn.datasets import fetch_openml
ames = fetch_openml(name = 'house_prices' , as_frame = True)
df = ames.frame
df = df.copy()
```

```python
df.head()
```

```python
#1. selecting the collumns
df.columns
```

```python
columns = [
    "OverallQual",
    "OverallCond",
    "GrLivArea",
    "CentralAir",
    "TotalBsmtSF",
    "SalePrice",
]

df = df[columns]
df.head()
```

```python
# 1. mapping the Y and N
df['CentralAir'] = df['CentralAir'].map({'N': 0, 'Y': 1})
df['CentralAir'].value_counts()
```

```python
df.isnull().sum()
```

CREATING THE Random forrest Regressor

```python
# 1. selecting the data
target = 'SalePrice'
features = df.columns[df.columns != target]

print(target)
print(features)
```

```python
X = df[features].values
y = df[target].values

from sklearn.model_selection import train_test_split
X_train, X_test, y_train, y_test = train_test_split(X,
                                                    y,
                                                    test_size = 0.3,
                                                    random_state = 42)


# 1. creating the RandomforestRegressor
from sklearn.ensemble import RandomForestRegressor

forest = RandomForestRegressor(n_estimators=100,
                               criterion='squared_error',
                               random_state= 42,
                               n_jobs = -1)

#2. fitting the data to the Algo
forest.fit(X_train , y_train)
y_train_pred = forest.predict(X_train)
y_test_pred = forest.predict(X_test)


# 3. finding the Eroor
from sklearn.metrics import mean_absolute_error , r2_score
mae_train = mean_absolute_error(y_train, y_train_pred)
mae_test = mean_absolute_error(y_test, y_test_pred)

print(f'MAE Train: {mae_train:.2f}')
print(f'MAE Test: {mae_test:.2f}')
print("-"*50)
print(f'R2 Score Train: {r2_score(y_train, y_train_pred):.2f}')
print(f'R2 Score Test: {r2_score(y_test, y_test_pred):.2f}')
```
